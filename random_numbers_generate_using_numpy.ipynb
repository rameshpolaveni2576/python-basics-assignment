random numbers generations using Numpy Practice

```python
import numpy as np
exe1=np.random.randint(1,100,5)
print(exe1)
```

```python
exe3=np.random.randint(1,100,(3,4))
print(exe3)
```

```python
exe2=np.random.randint(1,100,(5,5,5))
print(exe2)
```

Float value random generation

```python
exe4=np.random.random(5)
print(exe4)
```

```python
exe5=np.random.uniform(10,20,4)
print(exe5)
```

```python
np.random.choice(exe1,10)
```

```python
cities=np.array(["Delhi","Mumbai","Kolkatha","Hyderabad","Karimnagar","Warangal"])
cities_choice=np.random.choice(cities,2)
print(cities_choice)
```

```python
cities_choice_withoutduplicate=np.random.choice(cities,6,replace=False)
print(cities_choice_withoutduplicate)
```

Seed parameter

```python
print(np.array(["King","Queen","Jockey","ACE","Two","Three",
                "Four","Five","Six","Seven","Eight","Nine","Ten"]))
```

```python
deckofcards=np.array(["King","Queen","Jockey","ACE","Two","Three",
                "Four","Five","Six","Seven","Eight","Nine","Ten"])
np.random.seed(4)
np.random.shuffle(deckofcards)
print(deckofcards)
```
