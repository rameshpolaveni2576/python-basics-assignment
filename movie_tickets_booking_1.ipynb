```python
total_tickets=30
remaining_seats=total_tickets
tickets_sold=0
rejected_bookings=0
while remaining_seats>0:
  user_input=int(input("Enter the no of Tickets required? or 0 to exit: "))
  if user_input==0:
    print(f"Final Report is here:\nTotal tickets sold: {tickets_sold}\nrejeted bookings: {rejected_bookings}\nRemaining seats: {remaining_seats}")
    break
  if 1<user_input>15:
    print("---Invalid Booking Please enter valid no of tickets between 1-15:---")
  else:
    user_age=int(input("Enter your Age: "))
    if user_age<12:
      print("BOOKING REJECTED - Age restriction")
      rejected_bookings=rejected_bookings+1
    else:
      print("Booking Successful 😊")
      tickets_sold=tickets_sold+user_input
      remaining_seats=remaining_seats-user_input
      print(f"Tickets Sold: {tickets_sold}\nRemaining seats available to book: {remaining_seats}\nRejected Bookings: {rejected_bookings}")
if remaining_seats==0:
  print("----House Full-----")
```
