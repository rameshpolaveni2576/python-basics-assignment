```python
import json
Student={
    "name":"Ramesh",
    "age":"28",
    "Passed":True,
    "marks":[97,52,58],
    "address":{"city":"Hyderabad","pincode":"500038"}
    }
formatted_json_string=json.dumps(Student,indent=4)
print(formatted_json_string)
```

```python
to_python_dict=json.loads(formatted_json_string)
print(f"{to_python_dict["name"]}\n{to_python_dict["marks"][1]}\n{to_python_dict["address"]["city"]}")
```
