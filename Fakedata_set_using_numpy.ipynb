Fake data set

```python
import numpy as np
np.random.seed(69)
no_of_students=30
first_names=np.random.choice(["Ramesh","Vrushali","Anusha","Rivansh","Dinnu",
                              "Kiwi","Varamma","Odelu","Kumar","Sampath","Abhilash","Hardhik"],no_of_students)
print(first_names)
cities=np.random.choice(["Karimnagar","Rangapet","Manakondur","Veldi","Lalithapur",
                              "Devampally","Siddipet"],no_of_students)
print(cities)
ages=np.random.randint(18,26,no_of_students)
print(ages)
maths_marks=np.random.randint(35,100,no_of_students)
print(maths_marks)
physics_marks=np.random.randint(35,100,no_of_students)
print(physics_marks)
chemistry_marks=np.random.randint(35,100,no_of_students)
print(chemistry_marks)
eng_marks=np.random.randint(35,100,no_of_students)
print(eng_marks)
```

```python
with open("students_data.txt","w") as file:
  file.write("Name,City,Age,Maths,Physics,Chemistry,English\n")
  for i in range(no_of_students):
    file.write(f"{first_names[i]},{cities[i]},{ages[i]},{maths_marks[i]},{physics_marks[i]},{chemistry_marks[i]},{eng_marks[i]}\n")
```
