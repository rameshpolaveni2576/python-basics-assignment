```python
import seaborn as sns
titanic=sns.load_dataset("titanic")
titanic
```

```python
titanic["survived"]==1
```

```python
titanic[titanic["survived"]==1]
```

```python
titanic[titanic["class"]=="First"]
```

To use more filters here

```python
titanic[(titanic["class"]=="Third") & (titanic["survived"]==1)]
```

```python
titanic[(titanic["class"]=="Third") & (titanic["survived"]==1)&(titanic["sex"]=="female")]
```

```python
third_class_female_surviveors=titanic[(titanic["class"]=="Third") & (titanic["survived"]==1)&(titanic["sex"]=="female")]
print(f"3rd class fem sur are: {len(third_class_female_surviveors)}")
```

```python
age_less10_greatthan70=titanic[(titanic["age"]<10) | (titanic["age"]>70)]
print(f"age less than 10 and gtr than 70: {len(age_less10_greatthan70)}")
```

```python
is_in_use=titanic[(titanic["embarked"].isin(["S","Q"]))&(titanic["age"]<5)]
is_in_use
```

```python
is_in_use=titanic[(titanic["embarked"].isin(["S","Q"]))&(titanic["age"]<2)]
print(len(is_in_use))
```

```python
null_val=titanic[titanic["age"].isna()]
print(len(null_val))
```

```python
sorted_titanic=titanic.sort_values(["age"])
sorted_titanic
```

```python
#default is ascending order if we need descending need to use
sorted_titanic=titanic.sort_values(["age"],ascending=False)
sorted_titanic
```

```python
#sorting multiple coloumns is like below first it will sort the first value and then second like this
multiple_sorted_titanic=titanic.sort_values(["class","age"])
multiple_sorted_titanic
```

```python
#to control ascend and descend in this multiple coloumns
acdc_multiple_sorted_titanic=titanic.sort_values(["class","age"],ascending=[True,False])
acdc_multiple_sorted_titanic
```
