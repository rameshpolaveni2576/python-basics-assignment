```python
number_of_tickets=350
rejected_bookings=0
tickets_sold=0
while number_of_tickets<=350:
  user_input=int(input("Please Enter number of tickets here: "))
  if user_input>(350-tickets_sold) and tickets_sold!=350:
    print(f"Available Tickets: {number_of_tickets}")
    print(f"Please enter the tickets with in the range of available tickets")
  elif tickets_sold==350:
    print("-------House Full:All Tickets Booked Thank You--------")
  else:
    if user_input>15 or user_input==0:
      print("Please enter valid ticket number between 1-15 only")
    else:
      user_age=int(input("Please Enter your Age: "))
      if user_age>=12:
        print("Your booking is Successfull")
        number_of_tickets=number_of_tickets-user_input
        tickets_sold=tickets_sold+user_input
        print(f"Total Tickets Sold: {tickets_sold}")
        print(f"Rejected Bookings: {rejected_bookings}")
        print(f"Remaining Seats are: {number_of_tickets}")
      else:
        print("Invalid Booking Due to Age restriction")
        rejected_bookings=rejected_bookings+1
        continue

```
