```python
#TASK 1
import numpy as np
np.random.seed(42)
no_of_students=10
names=np.random.choice(["Ramesh","Anil","Anusha","Rivansh","Vrushali"],no_of_students)
age=np.random.randint(15,18,no_of_students)
maths=np.random.randint(40,100,no_of_students)
science=np.random.randint(40,100,no_of_students)
english=np.random.randint(40,100,no_of_students)
marks=np.array([maths,science,english]).T
#print(marks)
print(marks.shape)
print(marks.dtype)
```

```python
marks_reshaped=marks.reshape(2,5,3)
print(marks_reshaped)
marks_sliced=marks[0:5,0:2]
print(marks_sliced)
# Shape (2, 5, 3) means the marks are organized into 2 groups (blocks), each with 5 students (rows), and each student has 3 subject marks (columns). It’s a 3D NumPy array.
```

```python
with open("student_records.txt","w") as file:
  file.write("name | age | maths | science | english\n")
  for i in range(no_of_students):
    file.write(f"{names[i]} | {age[i]} | {maths[i]} | {science[i]} | {english[i]}\n")
  print("records written successfully\n")
with open("student_records.txt", "r") as file:
  print(file.read())
```
